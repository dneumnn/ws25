## Tokenizer

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
print(type(tokenizer))
```

| Tokenizer           | Implementation  | Backend                                                                                                                                        |
| ------------------- | --------------- | --------------------|
| `GPT2Tokenizer`     | **Pure Python** | Uses Hugging Face’s legacy Python tokenization logic                                                                                           |
| `GPT2TokenizerFast` | **Rust-based**  | Uses the high-performance [🤗 Tokenizers](https://github.com/huggingface/tokenizers) library written in Rust, then exposed via Python bindings |


## Vocabulary

```python
########## Lets have a look at the vocabulary of GPT-2 ##########
vocab = tokenizer.get_vocab() #key=subword, value=token_id
print(f"vocabulary is of type {type(vocab)} with size of {len(vocab)} tokens")
```

```python
s = 2000
for i, (token, token_id) in enumerate(vocab.items()): 
    if i > s:
        print(f"'{token}'", token_id) #, end=", ")
    if i > s+5: break
```

| Model       | Tokenizer Type               | Backend                                                       |
| ----------- | ---------------------------- | ------------------------------------------------------------- |
| **GPT-2**   | **Byte Pair Encoding (BPE)** | Hugging Face’s `ByteLevelBPETokenizer` (same as OpenAI GPT-2) |
| **LLaMA-2** | **SentencePiece (Unigram)**  | Meta’s `SentencePieceProcessor`                               |


A BPE tokenizer works directly on bytes, not Unicode characters. It breaks text into the most frequent byte-pairs.
Example: "fantastic" → ['fant', 'astic']
Uses special bytes for spaces and punctuation, e.g. Ġ for a space. So " hello" becomes ['Ġhello'].

# Tokenization

```python
prompt = "Deutschsprachige LLMs sind cool."
```

```python
input_ids = tokenizer.encode(prompt)
input_ids
```

```python
for token_id in input_ids:
    subword = tokenizer.decode(token_id)
    encode = tokenizer.encode(subword)
    print(token_id, "\t->", f"'{subword}' {encode}")
```

```python
output = tokenizer.decode(input_ids)
output
```

## TikToken

```python
# !pip install tiktoken
```

```python
import tiktoken
tiktoken.list_encoding_names()
```

```python
MODEL_ID = "gpt-4o"
tokenizer = tiktoken.encoding_for_model(MODEL_ID)
print("Model        :", MODEL_ID)
print("Encoding-Name:", tiktoken.encoding_name_for_model(MODEL_ID))
print("Vocabulary   :", tokenizer.n_vocab)
```

```python
input_ids = tokenizer.encode(prompt)
input_ids
```

```python
for token_id in input_ids:
    print(token_id, "\t", tokenizer.decode([token_id]))
```
